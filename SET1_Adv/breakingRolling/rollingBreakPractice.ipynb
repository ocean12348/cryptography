```python
plaintext = "Burning 'em, if you ain't quick and nimble\nI go crazy when I hear a cymbal"
key = "ICE"


keyBytes = key.encode()

pt = plaintext.encode()
ct = bytearray()

for i, byte in enumerate(pt):
    byteKey = keyBytes[i % len(keyBytes)]
    ct.append(byte ^ byteKey)

print(ct.hex())
```

```python
def hammingDist(str1:str, str2:str):
    if len(str1) != len(str2):
        raise ValueError("str not equal lenghts")
    bitstring1 = str1.encode()
    bitstring2 = str2.encode()

    distStr = bytearray()
    for a in bitstring1:
        for b in bitstring2:
            distStr.append(a^b)

    print(distStr.count(0b1))
```
